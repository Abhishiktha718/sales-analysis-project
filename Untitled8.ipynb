```python
import pandas as pd

df = pd.read_csv('Superstore.csv', encoding='latin1')
df.head()
```

```python
# remove duplicates
df = df.drop_duplicates()

# check missing values
df.isnull().sum()
```

```python
# drop missing values (simple method)
df = df.dropna()
```

```python
# convert date
df['Order Date'] = pd.to_datetime(df['Order Date'])

# create month column
df['Month'] = df['Order Date'].dt.month
```

```python
# sales by month
monthly_sales = df.groupby('Month')['Sales'].sum()

# profit by category
category_profit = df.groupby('Category')['Profit'].sum()

monthly_sales
category_profit
```

```python
import matplotlib.pyplot as plt

monthly_sales.plot()
plt.show()

category_profit.plot(kind='bar')
plt.show()
```

INSIGHTS :
The analysis shows that Technology is the most profitable category, while Furniture contributes minimal profit, indicating inefficiencies. Monthly sales trends reveal a clear seasonal pattern, with peak sales in Month 11 and lower performance in early months. Sales fluctuations across months suggest varying demand, highlighting opportunities for better planning and targeted strategies.
